```python
import pandas as pd 
import numpy as np
import re
```

```python
df = pd.read_csv('/Users/admin/Documents/ML Engineer/Practice/Data Cleaning/surat_uncleaned.csv')
df.head()
```

```python
df.info()
```

```python
df = df.rename(columns={
    'areaWithType': 'area_with_type'
})

df.head()
```

```python
def clean_prop_name(value):
    words = value.split() # value adalah string bukan series 

    # ambil angka jika kata pertama ada digit
    first = words[0]
    number = first if first.isdigit() else None

    #Gabungkan sisa kata (buang kata pertama jika digit)
    remaining = ' '.join(words[1:]) if number else ' '.join(words)

    #Split berdasarkan 'for Sale in'
    parts = remaining.split('for Sale in')
    prop_type = parts[0].strip() if len(parts) > 0 else None
    location = parts[1].strip() if len(parts) > 1 else None 

    return number, prop_type, location 

df[['number_prop', 'name_prop', 'location_prop']] = df['property_name'].apply(clean_prop_name).apply(pd.Series) # Cara paling aman tetap pakai .apply(pd.Series) setelah fungsi return tuple.
df.head()
```

```python
def clean_square_feet(value):
    match = re.search(r"(\d+)?\s*([a-zA-Z]+)", str(value))
    if match:
        return pd.Series([match.group(1), match.group(2)])
    return pd.Series([None, None])

df[['land_size', 'unit_size']] = df['square_feet'].apply(clean_square_feet)
df = df.drop('square_feet', axis=1)
```

```python
df.head()
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.replace("₹|,|per sqft", "", regex=True).str.strip()
df['price_per_sqft'] = pd.to_numeric(df['price_per_sqft'])
df.head()
```

```python
df['price'] = df['price'].str.replace("₹|Lac", "", regex=True).str.strip()
df = df.rename(columns={
    'price': 'price_lac'
})
df.head()
```

```python
df['floor'].value_counts()
```

```python
extracted = df['floor'].str.extract(f"^(.+?)\s+out of\s+(\d+)$")
df['floor_position'] = extracted[0]
df['total_floors'] = extracted[1]

mask = df['floor'].str.match(r"^Ground$", na=False)
df.loc[mask, 'floor_position'] = 'Ground'

status_values = ['Resale', 'New Property', 'Ready to Move', 'Const. Age New Construction']
df['status_property'] = np.where(df['floor'].isin(status_values), df['floor'], np.nan)

df['property_type_from_floor']  = np.where(
    df['floor'].isin(['Freehold']), df['floor'], np.nan
)

df['furnishing_from_floor'] = np.where(
    df['floor'].isin(['Unfurnished', 'Furnished', 'Semi-Furnished']), df['floor'], np.nan
)

df['building_name'] = np.where(
    df['floor'].isin(["The Polaris Avenue", "Om Residency", "Milestone Regalia"]), df['floor'], np.nan
)

df['society'] = np.where(
    df['floor'].isin(['Co-operative Society']), df['floor'], np.nan
)

sudah_masuk = df['floor'].isin(['Resale', 'New Property', 'Freehold', 'Ready to Move', 'Co-operative Society', 'Unfurnished', 'Furnished', 'Semi-Furnished', "The Polaris Avenue", "Om Residency", "Milestone Regalia"]) | df['floor'].str.match(r"^(.+?)\s+out of\s+(\d+)$", na=False) | df['floor'].str.match(r"^Ground$", na=False)
df['others'] = np.where(
    ~sudah_masuk, df['floor'], np.nan
)
```

```python
df.head()
```

```python
df['poss_date'] = df['status'].str.extract(r"Poss\. by (.+)")

df['status_property'] = np.where(df['status'].isin(['Resale', 'New Property', 'Ready to Move','Const. Age New Construction']), df['status'], np.nan)
df['property_type_from_floor'] = np.where(df['status'].isin(['Freehold']), df['status'], np.nan)
df['society'] = np.where(df['status'].isin(['Co-operative Society']), df['status'], np.nan)

df['measurement_size'] = df['status'].str.extract(r"(\d+\.?\d*\s*X\s*\d+\.?\d*)")
```
